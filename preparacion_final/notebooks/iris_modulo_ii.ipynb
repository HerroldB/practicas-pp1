# Análisis básico del conjunto Iris

Elegí este conjunto porque es pequeño y tiene medidas de flores y su especie. Primero cargué y revisé los datos. Después hice un gráfico para comparar el largo promedio del pétalo.

```python
# Primero importé las librerías que usé para leer la tabla y hacer el gráfico.
import pandas as pd
import matplotlib.pyplot as plt

# Después indiqué de dónde saqué los datos y qué nombre tiene cada columna.
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
columnas = [
    "largo_sepalo_cm",
    "ancho_sepalo_cm",
    "largo_petalo_cm",
    "ancho_petalo_cm",
    "especie"
]

# Leí el CSV y le puse los nombres de columnas porque el archivo no los trae.
datos = pd.read_csv(url, header=None, names=columnas)

# Mostré las primeras filas para revisar que se haya cargado bien.
print(datos.head())
```

```python
# Después revisé cuántas filas y columnas tenía la tabla.
print("Filas y columnas:", datos.shape)

# También revisé los tipos de datos y si había valores vacíos.
datos.info()

# Con describe vi un resumen de las columnas numéricas.
print(datos.describe().round(2))
```

```python
# Conté cuántos registros había de cada especie.
print("Registros por especie:")
print(datos["especie"].value_counts())

# Revisé si alguna columna tenía datos vacíos.
print("\nDatos faltantes por columna:")
print(datos.isna().sum())
```

```python
# Para hacer el gráfico, calculé el promedio del largo del pétalo por especie.
promedio_petalo = datos.groupby("especie")["largo_petalo_cm"].mean()

# Dibujé el promedio para poder comparar las tres especies.
plt.bar(promedio_petalo.index, promedio_petalo.values)
plt.title("Largo promedio del pétalo por especie")
plt.xlabel("Especie")
plt.ylabel("Largo del pétalo (cm)")
plt.xticks(rotation=15)
plt.tight_layout()

# Guardé el gráfico y también lo mostré en el notebook.
plt.savefig("grafico_largo_petalo.svg", bbox_inches="tight")
plt.show()
```
